```python
apps = apps[
    apps['Category'].isin(
        apps['Category'].dropna().unique()
    )
].copy()
```

```python
apps = apps[
    apps['Category'].str.match(
        r'^[A-Z_]+$', na=False
    )
].copy()
```

```python
apps['Type'] = apps['Type'].fillna('Free')
apps['Content Rating'] = apps['Content Rating'].fillna('Unknown')
apps['Current Ver'] = apps['Current Ver'].fillna('Unknown')
apps['Android Ver'] = apps['Android Ver'].fillna('Unknown')
```

```python
apps['Reviews'] = pd.to_numeric(
    apps['Reviews'],
    errors='coerce'
)
```

```python
apps['Installs'] = (
    apps['Installs']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False)
)

apps['Installs'] = pd.to_numeric(
    apps['Installs'],
    errors='coerce'
)
```

```python
apps['Price'] = (
    apps['Price']
    .astype(str)
    .str.replace('$', '', regex=False)
)

apps['Price'] = pd.to_numeric(
    apps['Price'],
    errors='coerce'
)
```

```python
def convert_size(size):
    size = str(size)

    if size == 'Varies with device':
        return np.nan

    if size.endswith('M'):
        return float(size.replace('M', ''))

    if size.endswith('k'):
        return float(size.replace('k', '')) / 1024

    return np.nan

apps['Size_MB'] = apps['Size'].apply(convert_size)
```

```python
apps['Last Updated'] = pd.to_datetime(
    apps['Last Updated'],
    errors='coerce'
)
```

```python
reviews = reviews.dropna(
    subset=['App', 'Translated_Review']
)

reviews = reviews.drop_duplicates(
    subset=['App', 'Translated_Review']
)

print("Review cleaning completed!")
```

```python
!pip install vaderSentiment
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()
```

```python
def sentiment_analysis(text):

    score = analyzer.polarity_scores(str(text))

    compound = score['compound']

    if compound >= 0.05:
        return 'Positive'

    elif compound <= -0.05:
        return 'Negative'

    else:
        return 'Neutral'
```

```python
reviews['VADER_Sentiment'] = (
    reviews['Translated_Review']
    .apply(sentiment_analysis)
)
```

```python
reviews[
    ['App', 'Translated_Review', 'VADER_Sentiment']
].head(10)
```

```python
sentiment_count = reviews[
    'VADER_Sentiment'
].value_counts()

plt.figure(figsize=(8,5))

sns.barplot(
    x=sentiment_count.index,
    y=sentiment_count.values
)

plt.title('User Review Sentiment Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Number of Reviews')

plt.show()
```

```python
merged_data = pd.merge(
    reviews,
    apps[['App', 'Category']],
    on='App',
    how='inner'
)

merged_data.head()
```

```python
category_sentiment = pd.crosstab(
    merged_data['Category'],
    merged_data['VADER_Sentiment']
)

category_sentiment
```

```python
category_sentiment.plot(
    kind='bar',
    figsize=(15,7)
)

plt.title('Sentiment Analysis by App Category')
plt.xlabel('App Category')
plt.ylabel('Number of Reviews')

plt.xticks(rotation=90)
plt.legend(title='Sentiment')

plt.show()
```

```python
positive_reviews = (
    merged_data[
        merged_data['VADER_Sentiment'] == 'Positive'
    ]
    .groupby('Category')
    .size()
    .sort_values(ascending=False)
)

negative_reviews = (
    merged_data[
        merged_data['VADER_Sentiment'] == 'Negative'
    ]
    .groupby('Category')
    .size()
    .sort_values(ascending=False)
)

print("Top positive categories:")
print(positive_reviews.head(5))

print("\nTop negative categories:")
print(negative_reviews.head(5))
```

```python
sentiment_percentage = pd.crosstab(
    merged_data['Category'],
    merged_data['VADER_Sentiment'],
    normalize='index'
) * 100

sentiment_percentage
```
